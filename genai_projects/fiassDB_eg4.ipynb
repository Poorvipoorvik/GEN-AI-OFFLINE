```python
from langchain_core.documents import Document

from langchain_community.vectorstores.faiss import FAISS
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

## step 1 : take paragraph from chatgpt

```python
documents = [

    Document(
        page_content="""
        The Magic Garden

        Once upon a time, a little girl named Lily discovered
        a beautiful garden behind her grandmother's house.
        The garden was filled with colorful flowers and talking
        butterflies. One evening, she found a tiny golden door
        hidden beneath a rose plant. When she opened it, she
        discovered a magical world where flowers could grant wishes.
        Lily wished for happiness and good health for everyone
        in her village.
        """,
        metadata={"story": "The Magic Garden", "category": "Fantasy"}
    ),

    Document(
        page_content="""
        The Brave Little Dog

        A young boy named Sam had a pet dog called Bruno.
        One rainy evening, Sam got lost in a forest while playing.
        Bruno searched everywhere for him. After several hours,
        Bruno found Sam sitting near a large tree. The dog guided
        him safely back to the village. Sam hugged Bruno and
        realized that true friendship comes from love and loyalty.
        """,
        metadata={"story": "The Brave Little Dog", "category": "Adventure"}
    ),

    Document(
        page_content="""
        The Mystery of the Old House

        A young detective named Alex moved into a new town.
        Near his house, there was an abandoned mansion that nobody
        visited. One night, Alex noticed a light shining through
        one of its windows. He entered the mansion and discovered
        an old diary hidden inside a wooden cupboard. The diary
        revealed the secret history of the mansion and its missing
        owner. Alex decided to investigate the mystery.
        """,
        metadata={"story": "The Mystery of the Old House", "category": "Mystery"}
    ),

    Document(
        page_content="""
        The Journey to Mars

        In the year 2050, a scientist named David travelled to Mars
        with a team of astronauts. Their mission was to discover
        whether humans could live on the planet. During their
        exploration, they discovered a strange underground lake
        containing frozen water. The discovery gave scientists
        hope that humans might build a future colony on Mars.
        David returned to Earth as a famous explorer.
        """,
        metadata={"story": "The Journey to Mars", "category": "Science Fiction"}
    ),

    Document(
        page_content="""
        The Honest Farmer

        There was a poor farmer named Ramesh who lived in a small
        village. One morning, while working in his field, he found
        a bag full of gold coins. Instead of keeping the money,
        he searched for its owner. Later, he discovered that the
        bag belonged to a wealthy merchant. The merchant was
        impressed by Ramesh's honesty and rewarded him with enough
        money to improve his farm. Ramesh continued to live an
        honest and peaceful life.
        """,
        metadata={"story": "The Honest Farmer", "category": "Moral Story"}
    )

]
```

## step 2 : creating embedding model

```python
model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-mpnet-base-v2"
)
```

## Step 3 : create FAISS DB.

```python
'''
    while create faiss db , give chunks and embedding model
'''
```

```python
# Create FAISS Vector database
vector_db=FAISS.from_documents(
    documents=documents,
    embedding=model
)
```

## Step 4 : ask a question to faiss db based on the chunks

```python
results = vector_db.similarity_search('who is the honest farmer',k = 1)
```

```python
results[0].page_content
```
